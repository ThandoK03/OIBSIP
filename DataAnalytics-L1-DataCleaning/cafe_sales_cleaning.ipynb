```python
import pandas as pd
import numpy as np

df = pd.read_csv('cafe_sales_dirty.csv')
df.head()
```

## Step 1: Data Quality Report

Before cleaning, I inspect the dataset's shape, data types, and the extent of missing/invalid values — including hidden nulls like "ERROR" and "UNKNOWN" strings that don't show up as standard NaN.

```python
# Shape and structure
print("Shape:", df.shape)
print("\nData types:")
print(df.dtypes)

# Standard nulls
print("\nMissing values (NaN) per column:")
print(df.isnull().sum())

# Hidden nulls - ERROR and UNKNOWN as strings
print("\n'ERROR' values per column:")
print((df == 'ERROR').sum())

print("\n'UNKNOWN' values per column:")
print((df == 'UNKNOWN').sum())

# Duplicates
print("\nDuplicate rows:", df.duplicated().sum())
```
